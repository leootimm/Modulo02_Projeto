```python
import pandas as pd

# leitura base
df=pd.read_csv('acidentes2025.csv', sep=';', encoding='latin1')

# cabeçalho
print(' :: ESTATÍSTICAS DE ACIDENTES EM ESTRADAS FEDERAIS')
print(' :: ANO 2025')
print()

# estatisticas iniciais
print(f"No ano de 2025 ocorreram {df.shape[0]:,} acidentes nas estradas federais brasileiras.".replace(',', '.'))

dia = df[df['fase_dia'].isin(['Pleno dia', 'Amanhecer'])].shape[0]
noite = df[df['fase_dia'].isin(['Plena Noite', 'Anoitecer'])].shape[0]
print('Destes,',f"{dia:,} foram durante o dia e {noite:,} a noite.".replace(',', '.'))

obitos = df[df['estado_fisico'] == 'Óbito']
obitos['fase_dia'].value_counts()
dia = obitos[obitos['fase_dia'].isin(['Pleno dia', 'Amanhecer'])].shape[0]
noite = obitos[obitos['fase_dia'].isin(['Plena Noite', 'Anoitecer'])].shape[0]
print(f"{noite:,} acidentes a noite e {dia:,} de dia tiveram óbito.".replace(',', '.'))

# demais consultas
print('_____________________________________')
print()

# criacao da coluna Faixa de Horário
horas = df['horario'].str[:2].astype(int)
limites = [-1, 5, 11, 17, 23]
rotulos = ['Madrugada', 'Manhã', 'Tarde', 'Noite']
df['faixa_horario'] = pd.cut(horas, bins=limites, labels=rotulos)
print('Em qual período do dia ocorreram mais acidentes?')
display((df['faixa_horario'].value_counts(normalize=True) * 100).map('{:.2f}%'.format).rename_axis(None).to_frame(''))

print('_____________________________________')
print()

print(f"Qual o tipo de pista?")
display(df['tipo_pista'].value_counts().rename_axis(None).to_frame(''))
print('_____________________________________')
print()

print('Principais causas de acidentes:')
display(df['causa_acidente'].value_counts().head(5).rename_axis(None).to_frame(''))
print('_____________________________________')
print()

meses_nome = {'01': 'Janeiro', '02': 'Fevereiro', '03': 'Março', '04': 'Abril', '05': 'Maio', '06': 'Junho',
              '07': 'Julho', '08': 'Agosto', '09': 'Setembro', '10': 'Outubro', '11': 'Novembro', '12': 'Dezembro'}
print('Em 2025, quais foram  os meses do ano com mais acidentes:')
display(df['data_inversa'].str[3:5].map(meses_nome).value_counts().head(3).rename_axis(None).to_frame(''))

print('_____________________________________')
print()

print('Qual o tipo de acidente possui mais vitimas fatais?')
fatais = df[df['classificacao_acidente'] == 'Com Vítimas Fatais']
tipo_mais_fatal = fatais['tipo_acidente'].value_counts().idxmax()
qtd_fatais = fatais['tipo_acidente'].value_counts().max()

print(f"- {tipo_mais_fatal} com {qtd_fatais} ocorrências.")

print('_____________________________________')
print()

print('Quais as 3 estradas federais possuem mais acidentes:')
display(df["br"].value_counts().head(3).to_frame(name='').rename_axis("BR"))

print('_____________________________________')
print()

print('Quais são os 10 municípios cujos trechos de rodovias federais registram mais acidentes?')
top_10_cidades = df[['municipio']].value_counts().head(10)
display(top_10_cidades.rename_axis(None).to_frame(''))
print('_____________________________________')
print()


print('Percentual de informações em branco em comparação ao total de registros:')
nulos = (df.isnull().sum() / df.shape[0] * 100).round(2)
display(nulos[nulos > 0].map('{:.2f}%'.format).rename_axis(None).to_frame(''))
```
